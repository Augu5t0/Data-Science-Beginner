```python
import pandas as pd
```

```python
dados_condominio = pd.read_json('/content/drive/MyDrive/Dados_Json/dados_locacao_imoveis.json')
```

```python
dados_condominio
```

```python
dados_condominio = pd.json_normalize(dados_condominio['dados_locacao'])
dados_condominio
```

```python
dados_condominio = dados_condominio.explode(
    ['valor_aluguel', 'datas_combinadas_pagamento', 'datas_de_pagamento']
).reset_index(drop=True)
```

```python
dados_condominio
```

```python
dados_condominio.info()
```

```python
dados_condominio['valor_aluguel'] = dados_condominio['valor_aluguel'].astype(str).replace(r'[\$\ reais]', '', regex=True).\
                                              replace(r',', '.', regex=True).astype(float)
```

```python
dados_condominio.info()
```

```python
dados_condominio
```

```python
dados_condominio['datas_de_pagamento'] = pd.to_datetime(dados_condominio['datas_de_pagamento'], format='%d/%m/%Y')
dados_condominio['datas_combinadas_pagamento'] = pd.to_datetime(dados_condominio['datas_combinadas_pagamento'], format='%d/%m/%Y')
```

```python
dados_condominio
```

```python
dados_condominio.info()
```

```python
dados_condominio['apartamento'] = dados_condominio['apartamento'].str.replace(r'\s*\(\s*blocoAP\s*\)', '', regex=True)
```
