#### Normalização

como os dados estão aninhados, nos precisamos destrinchar eles, por isso vamos aplicar a normalização nesses dados

```python
dados = pd.json_normalize(dados['info_moveis'])
dados
```

Existem colunas que estão agrupados em lista e para desagrupar esses dados nos iremos usar o métdodo Explode

```python
colunas =  list(dados.columns)
colunas
```

```python
dados = dados.explode(colunas[3:])
dados.reset_index(inplace=True, drop=True)
dados
```

```python
dados.info()
```

#### Dados Textuais

Frequentemente um dado textual pode ser usado como parametro para fazer um estudo de imovel, pensando nisso é positivo tratar esses dados textuais

##### Tokenização Simples

```python
dados.info()
```

```python
#transforma todos os caracteres em letras minúsculas
dados['descricao_local'] = dados['descricao_local'].str.lower()
```

```python
dados['descricao_local'][3169]
```

```python
# sobrescreve todo caracter que não é de A a Z, de a a z e de 0 a 9, além do - e do ' por um espaço
dados['descricao_local'] = dados['descricao_local'].str.replace(r'[^a-zA-Z0-9\-\']', ' ', regex=True)
```

```python
dados['descricao_local'] = dados['descricao_local'].str.replace(r'(?<!\w)-(?!\w)', ' ', regex=True)
```

```python
dados['descricao_local'][3169]
```

```python
#separa as palavras de acordo com o espaço em branco
dados['descricao_local'] = dados['descricao_local'].str.split()
```

```python
dados['descricao_local']
```

```python
# sobrescreve {} e "" por espaço em branco
dados['comodidades'] = dados['comodidades'].str.replace(r'\{|\}\"|\"', '', regex=True).str.split(',')
```

```python
dados['comodidades']
```

```python
#transforma todos os caracteres em letras minúsculas
dados['descricao_vizinhanca'] = dados['descricao_vizinhanca'].str.lower()
```

```python
 # sobrescreve todo caracter que não é de A a Z, de a a z e de 0 a 9, além do - e do ' por um espaço
 dados['descricao_vizinhanca'] = dados['descricao_vizinhanca'].str.replace(r'[^a-zA-Z0-9\-\']', ' ', regex=True).str.split()
```

```python
dados['descricao_vizinhanca']
```
