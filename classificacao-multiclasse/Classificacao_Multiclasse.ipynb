## Realizando treinamento

### Preparando os dados

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
colunas_categoricas = ['Migração', 'Sexo','Estrangeiro', 'Necessidades educacionais especiais', 'Devedor',
                     'Taxas de matrícula em dia', 'Bolsista', 'Período','Estado civil', 'Curso', 'Qualificação prévia']
```

```python
encoder = OneHotEncoder(drop='if_binary')

df_categorico = df[colunas_categoricas]

df_encoded = pd.DataFrame(encoder.fit_transform(df_categorico).toarray(),
                          columns=encoder.get_feature_names_out(colunas_categoricas))

df_final = pd.concat([df.drop(columns = colunas_categoricas), df_encoded],axis=1)
```

```python
df_final
```

### Separando os dados

```python
X = df_final.drop(columns=['Target'])
y = df_final['Target']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X, X_teste, y, y_teste = train_test_split(X, y, test_size=0.15, stratify=y, random_state=0)
X_treino, X_val, y_treino, y_val = train_test_split(X, y, stratify=y, random_state=0)
```

## Treinando o modelo

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
modelo_rf = RandomForestClassifier(random_state=0)
modelo_rf.fit(X_treino, y_treino)
y_pred = modelo_rf.predict(X_val)
print(f'Acuracia: {modelo_rf.score(X_treino, y_treino)}')
print(f'Acuracia: {modelo_rf.score(X_val, y_val)}')
```

```python
modelo_rf = RandomForestClassifier(random_state=0, max_depth=10)
modelo_rf.fit(X_treino, y_treino)
y_pred = modelo_rf.predict(X_val)
print(f'Acuracia: {modelo_rf.score(X_treino, y_treino)}')
print(f'Acuracia: {modelo_rf.score(X_val, y_val)}')
```

```python
from sklearn.ensemble import GradientBoostingClassifier
```

```python
modelo_gb = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=0)
modelo_gb.fit(X_treino, y_treino)
y_pred_gb = modelo_gb.predict(X_val)
print(f'Acuracia: {modelo_gb.score(X_treino, y_treino)}')
print(f'Acuracia: {modelo_gb.score(X_val, y_val)}')
```

## Avaliando os modelos

### Matriz de confusão

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
```

```python
matriz_confusao = confusion_matrix(y_val,y_pred)

visualizacao = ConfusionMatrixDisplay(confusion_matrix=matriz_confusao, display_labels=modelo_rf.classes_)
visualizacao.plot();
```

```python
ConfusionMatrixDisplay.from_predictions(y_val, y_pred, normalize='true', cmap='Blues');
```

```python
matriz_confusao = confusion_matrix(y_val,y_pred_gb)

visualizacao = ConfusionMatrixDisplay(confusion_matrix=matriz_confusao, display_labels=modelo_rf.classes_)
visualizacao.plot();
```

```python
ConfusionMatrixDisplay.from_predictions(y_val, y_pred_gb, normalize='true', cmap='Blues');
```

### Relátorio de métricas

```python
from sklearn.metrics import classification_report
```

```python
report = classification_report(y_val, y_pred, target_names=modelo_rf.classes_)
print('Relátorio de Classificação:\n', report)
```

```python
report = classification_report(y_val, y_pred_gb, target_names=modelo_rf.classes_)
print('Relátorio de Classificação:\n', report)
```

### Superamostragem (Oversampling)

```python
y_treino.value_counts()
```

```python
from imblearn.over_sampling import SMOTE
```

```python
oversample = SMOTE(random_state=0)
X_balanceado, y_balanceado = oversample.fit_resample(X_treino, y_treino)
```

```python
y_balanceado.value_counts()
```

```python
modelo = RandomForestClassifier(random_state=0, max_depth = 10)
modelo.fit(X_balanceado, y_balanceado)
y_pred_b = modelo.predict(X_val)
```

```python
modelo_gb = GradientBoostingClassifier(n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=0)
modelo_gb.fit(X_balanceado, y_balanceado)
y_pred_gb_b = modelo_gb.predict(X_val)
```

### Analisando as métricas após o balanceamento

```python
print(classification_report(y_val, y_pred_b, target_names=modelo.classes_))
print(ConfusionMatrixDisplay.from_predictions(y_val, y_pred_b, normalize='true', cmap='Blues'))
```

```python
print(classification_report(y_val, y_pred_gb_b, target_names=modelo.classes_))
print(ConfusionMatrixDisplay.from_predictions(y_val, y_pred_gb_b, normalize='true', cmap='Blues'))
```

## Validando Modelo

### Pipeline para validação

```python
from imblearn.pipeline import Pipeline as imbpipeline
```

```python
modelo = RandomForestClassifier(random_state=0, max_depth = 10)
pipeline = imbpipeline([('Oversample', SMOTE()),('Floresta', modelo)])
```

```python
from sklearn.model_selection import cross_validate, StratifiedKFold
```

```python
skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)
cv_resultado = cross_validate(pipeline, X, y, cv=skf, scoring='recall_weighted')
cv_resultado['test_score']
```

### Intervalo de Confiança

```python
media =  cv_resultado['test_score'].mean()
desvio_padrao = cv_resultado['test_score'].std()
print(f'Intervalo de Confiança: [{media - (2*desvio_padrao)}, {min(media + (2*desvio_padrao), 1.0)}]')
```

### Testando o modelo

```python
oversample = SMOTE(random_state=0)
X_balanceado, y_balanceado = oversample.fit_resample(X, y)
modelo = RandomForestClassifier(random_state=0, max_depth = 10)
modelo.fit(X_balanceado, y_balanceado)
y_pred = modelo.predict(X_teste)

print(classification_report(y_teste, y_pred))
ConfusionMatrixDisplay.from_predictions(y_teste, y_pred, normalize='true', cmap='Blues');
```
